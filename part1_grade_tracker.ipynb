# Task 1 — Data Parsing & Profile Cleaning

The data we have is kind of all over the place - names have random spaces and casing, roll numbers are strings, marks are just one long comma separated string. Going to clean it up and store everything in a proper list.

```python
# input data from the question

raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

raw_students[0]
```

```python
# trying to split the marks string
test = raw_students[0]["marks_str"].split(",")
test
```

```python
# there's a space after each comma. need to split on ", " instead

cleandata_students = []

for student in raw_students:
    clean_name = student["name"].strip().title()
    clean_roll = int(student["roll"])

    marks_list = []
    for m in student["marks_str"].split(", "):
        marks_list.append(int(m))

    cleandata_students.append({
        "name": clean_name,
        "roll": clean_roll,
        "marks": marks_list
    })

cleandata_students[0]
```

```python
for std in cleandata_students:
    words = std["name"].split()
    name_ok = True
    for w in words:
        if not w.isalpha():
            name_ok = False
            break

    tag = "Valid name" if name_ok else "Invalid name"

    print(f"================================")
    print(f"Student : {std['name']}  ({tag})")
    print(f"Roll No : {std['roll']}")
    print(f"Marks   : {std['marks']}")
    print(f"================================\n")
```

# Task 2 — Marks Analysis Using Loops & Conditionals

Report card for Ayesha - grades, highest/lowest, and a while loop to let the user add more subjects.

```python
def infer_grade(m):
    if m >= 90:
        return "A+"
    elif m >= 80:
        return "A"
    elif m >= 70:
        return "B"
    elif m >= 60:
        return "C"
    else:
        return "F"

infer_grade(88)
```

```python
student_name = "Ayesha Sharma"
subjects     = ["Math", "Physics", "CS", "English", "Chemistry"]
marks        = [88, 72, 95, 60, 78]

print(f"Report Card for {student_name}")
print("----------------------------------")

for i in range(len(subjects)):
    grade = infer_grade(marks[i])
    print(f"{subjects[i]:20s} : {marks[i]:3d}  Grade: {grade}")

print("------------------------------------------")

total = sum(marks)
average = round(total / len(marks), 2)

# find highest and lowest - start from first, compare forward
hi = 0
lo = 0
for i in range(1, len(marks)):
    if marks[i] > marks[hi]:
        hi = i
    if marks[i] < marks[lo]:
        lo = i

print(f"Total Marks          : {total}")
print(f"Average Marks        : {average}")
print(f"Highest Scoring      : {subjects[hi]} ({marks[hi]})")
print(f"Lowest Scoring       : {subjects[lo]} ({marks[lo]})")
print("------------------------------------------")
```

```python
print("=== Add New Subjects ===")
print("(type 'Quit' as subject name to end)\n")

new_count = 0

while True:
    sub = input("Enter subject name: ").strip()

    if sub.lower() == "quit":
        break
    if sub == "":
        print("  Subject name can't be empty, try again.\n")
        continue

    raw_mark = input(f"Enter marks for {sub} (0-100): ").strip()

    try:
        m = int(raw_mark)
    except ValueError:
        print(f"  That's not a valid number, skipping.\n")
        continue

    if m < 0 or m > 100:
        print(f"  Marks must be between 0 and 100.\n")
        continue

    subjects.append(sub)
    marks.append(m)
    new_count += 1
    print(f"  Added {sub} with {m} marks.\n")

print("----------------------------------")
print(f"New subjects added: {new_count}")

# updated avg
updated_total = 0
for m in marks:
    updated_total += m
print(f"Updated average (all {len(marks)} subjects): {round(updated_total / len(marks), 2)}")
```

# Task 3 — Class Performance Summary

```python
# using the cleaned data from task 1
class_data = [(s["name"], s["marks"]) for s in cleandata_students]

res = []
for name, marks in class_data:
    avg = round(sum(marks) / len(marks), 2)
    status = "Pass" if avg >= 60 else "Fail"
    res.append((name, avg, status))

res
```

```python
print(f"{'Name':>20}| {'Average':>7} | Status")
print("--------------------------------------------")
for name, avg, status in res:
    print(f"{name:>20}| {avg:>7}  | {status}")
print("--------------------------------------------")

# topper
topper = res[0]
for r in res:
    if r[1] > topper[1]:
        topper = r

pass_count = 0
fail_count = 0
for _, _, status in res:
    if status == "Pass":
        pass_count += 1
    else:
        fail_count += 1

class_avg = round(sum(r[1] for r in res) / len(res), 2)

print()
print(f"Passed : {pass_count}")
print(f"Failed : {fail_count}")
print(f"Topper : {topper[0]} ({topper[1]})")
print(f"Class Average : {class_avg}")
```

# Task 4 - String Manipulation Utility

strip, title case, count/replace 'python', split into sentences. careful with the sentence splitting - need to split on ". " not just "." 

```python
essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

clean = essay.strip()
print("Stripped:")
print(clean)
print()

print("Title Case:")
print(clean.title())
print()

count = clean.count("python")
print(f"'python' appears {count} time(s)")
print()

replaced = clean.replace("python", "Python 🐍")
print("After replacing:")
print(replaced)
```

```python
sentences = clean.split(". ")
for i in range(len(sentences)):
    print(f"  {i+1}. {sentences[i]}")
```

```python
# the first two lost their periods from the split, need to add them back
for i in range(len(sentences)):
    s = sentences[i]
    if not s.endswith("."):
        s = s + "."
    print(f"  {i+1}. {s}")
```
